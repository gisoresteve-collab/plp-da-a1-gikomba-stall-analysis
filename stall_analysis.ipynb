## Task 1: Overall Sales Performance

I will use Python's built-in functions to calculate total,
average, highest and lowest sales. I will then use the
position of the highest and lowest values to identify
the corresponding day names and weeks.

```python

# Calculate overall sales performance

total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Identify the best-performing day
best_day_index = daily_sales.index(highest_sales)
best_day = days[best_day_index]
best_week = (best_day_index // 7) + 1

# Identify the lowest-performing day
worst_day_index = daily_sales.index(lowest_sales)
worst_day = days[worst_day_index]
worst_week = (worst_day_index // 7) + 1

# Display the results
print("OVERALL SALES PERFORMANCE")
print("-------------------------")
print(f"Total sales:       KES {total_sales:,.2f}")
print(f"Average daily:     KES {average_sales:,.2f}")
print(f"Highest sales:     KES {highest_sales:,.2f}")
print(f"Lowest sales:      KES {lowest_sales:,.2f}")
print(f"Best day:          {best_day}, Week {best_week}")
print(f"Worst day:         {worst_day}, Week {worst_week}")
```

```python

# Task 2: Weekly Sales and Growth

def week_totals(sales):
    """Return the total sales for Week 1 and Week 2."""
    week1 = sum(sales[:7])
    week2 = sum(sales[7:14])
    return week1, week2

week1_total, week2_total = week_totals(daily_sales)

growth_percentage = ((week2_total - week1_total) / week1_total) * 100

print("WEEKLY SALES PERFORMANCE")
print(f"Week 1: KES {week1_total:,.2f}")
print(f"Week 2: KES {week2_total:,.2f}")
print(f"Week-on-week growth: {growth_percentage:.2f}%")
```

```python

# Task 3: Daily Sales Classification

def classify_day(amount, average):
    """Classify daily sales as Strong, Normal or Weak."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"

# Create a dictionary to count each category
category_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Classify every day and count the results
for amount in daily_sales:
    category = classify_day(amount, average_sales)
    category_counts[category] += 1

print("DAILY SALES CLASSIFICATION")

for category, count in category_counts.items():
    print(f"{category}: {count} days")
```

```python

# Task 4: Weekday vs Weekend Performance

weekday_sales = daily_sales[:5] + daily_sales[7:12]
weekend_sales = daily_sales[5:7] + daily_sales[12:14]

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

print("WEEKDAY VS WEEKEND PERFORMANCE")
print(f"Weekday average: KES {weekday_average:,.2f}")
print(f"Weekend average: KES {weekend_average:,.2f}")
print(f"Weekend increase: {percentage_difference:.2f}%")
```

```python

# Task 5: Inventory and Gross Margins

def products_to_reorder(inventory):
    """Return products whose stock is at or below reorder level."""
    reorder_list = []

    for product, details in inventory.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


# Identify products requiring restocking
reorder_products = products_to_reorder(stock)

print("INVENTORY AND GROSS MARGINS")

for product, details in stock.items():
    margin = (
        (details["price"] - details["cost"]) / details["price"]
    ) * 100

    print(f"{product}: Gross margin = {margin:.2f}%")

print("\nPRODUCTS REQUIRING RESTOCKING")

for product in reorder_products:
    print(product)
```

```python

# Task 6: Stall Performance Report

print("=" * 50)
print("          STALL PERFORMANCE REPORT")
print("             Gikomba Market")
print("=" * 50)

print("\n1. OVERALL SALES")
print(f"Total sales:          KES {total_sales:,.2f}")
print(f"Average daily sales:  KES {average_sales:,.2f}")
print(f"Highest sales:        KES {highest_sales:,.2f}")
print(f"Lowest sales:         KES {lowest_sales:,.2f}")
print(f"Best day:             {best_day}, Week {best_week}")
print(f"Worst day:            {worst_day}, Week {worst_week}")

print("\n2. WEEKLY PERFORMANCE")
print(f"Week 1 sales:         KES {week1_total:,.2f}")
print(f"Week 2 sales:         KES {week2_total:,.2f}")
print(f"Sales growth:         {growth_percentage:.2f}%")

print("\n3. DAILY CLASSIFICATION")
for category, count in category_counts.items():
    print(f"{category:<15}: {count} days")

print("\n4. WEEKDAY VS WEEKEND")
print(f"Weekday average:      KES {weekday_average:,.2f}")
print(f"Weekend average:      KES {weekend_average:,.2f}")
print(f"Weekend increase:     {percentage_difference:.2f}%")

print("\n5. INVENTORY ALERTS")
for product in reorder_products:
    print(f"RESTOCK: {product}")

print("\n6. GROSS MARGINS")
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) /
              details["price"]) * 100
    print(f"{product:<15}: {margin:.2f}%")

print("=" * 50)
print("             END OF REPORT")
print("=" * 50)
```
